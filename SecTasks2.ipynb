```python
import pandas as pd

df = pd.read_csv("amazon_mobiles.csv")
df.head()
```

```python
df.shape
df.columns
df.info()
```

```python
#Data Cleaning
df.isnull().sum()
df.dropna(inplace=True)
```

```python
df['price'] = df['price'].replace('[₹,]', '', regex=True).astype(float)
```

```python
df.describe()
df.columns
df['brand'] = df['product'].apply(lambda x: x.split()[0])
```

```python
#Brand analysis
df['brand'].value_counts()
```

```python
#top Rated Mobiles
df.sort_values(by='rating', ascending=False).head(10)
```

```python
#Visualization
import matplotlib.pyplot as plt
df['price'].hist()
plt.title("Price Distribution")
plt.show()
df['brand'].value_counts().head(10).plot(kind='bar')
plt.title("Top Brands")
plt.show()
```

```python
#Correlation if exists
df[['price', 'rating']].corr()
```

```python
#Data Visualizing
df['price'].hist()
plt.title("Price Distribution of Mobiles")
plt.xlabel("Price")
plt.ylabel("Count")
plt.show()
```

```python
df['brand'].value_counts().head(10).plot(kind='bar')
plt.title("Top 10 Mobile Brands")
plt.xlabel("Brand")
plt.ylabel("Count")
plt.show()
```

```python
df['rating'].hist()
plt.title("Rating Distribution")
plt.xlabel("Rating")
plt.ylabel("Count")
plt.show()
```

```python
plt.scatter(df['price'], df['rating'])
plt.title("Price vs Rating")
plt.xlabel("Price")
plt.ylabel("Rating")
plt.show()
```

```python
plt.boxplot(df['price'])
plt.title("Price Outliers")
plt.show()
```
